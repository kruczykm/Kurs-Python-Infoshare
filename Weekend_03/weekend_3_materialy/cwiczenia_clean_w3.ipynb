## Zadanie 5 — Walidacja całego zamówienia

**Opis biznesowy:** Rekord zamówienia jest poprawny tylko wtedy, gdy ma poprawną ilość i poprawną cenę.

**Cel techniczny:** Utworzyć funkcję `is_valid_order`, która korzysta z wcześniejszych walidatorów.

**Poziom trudności:** 2/5

**Szacowany czas:** 15 min

**Wskazówki:**

- Dane w `orders_raw` mają ilość i cenę zapisane jako tekst.
- W środku funkcji wykonaj konwersję do `int` i `float`.

```python
orders_raw = [
    {"invoice_no": "536365", "stock_code": "85123A", "description": "WHITE HANGING HEART T-LIGHT HOLDER", "quantity": "6", "unit_price": "2.55", "customer_id": "17850", "country": "United Kingdom"},
    {"invoice_no": "536366", "stock_code": "71053", "description": "WHITE METAL LANTERN", "quantity": "-1", "unit_price": "3.39", "customer_id": "17850", "country": "United Kingdom"},
    {"invoice_no": "536367", "stock_code": "84406B", "description": "CREAM CUPID HEARTS COAT HANGER", "quantity": "8", "unit_price": "0", "customer_id": "17850", "country": "France"},
    {"invoice_no": "536368", "stock_code": "84029G", "description": "KNITTED UNION FLAG HOT WATER BOTTLE", "quantity": "3", "unit_price": "3.39", "customer_id": None, "country": "Germany"},
    {"invoice_no": "536369", "stock_code": "22752", "description": "SET 7 BABUSHKA NESTING BOXES", "quantity": "2", "unit_price": "7.65", "customer_id": "13047", "country": "United Kingdom"}
]

def is_valid_quantity(quantity):
    return quantity > 0

def is_valid_price(price: float) -> bool:
     return price > 0

def is_valid_order(order: dict) -> bool:
     quantity = int(order["quantity"])
     price = float(order["unit_price"])

     return is_valid_quantity(quantity) and is_valid_price(price)


for order in orders_raw:
    print(order["invoice_no"], is_valid_order(order))
```

## Zadanie 6 — Czytelne nazwy zmiennych

**Opis biznesowy:** Kod raportu powinien używać pojęć biznesowych, a nie skrótów typu `s`, `c`, `bad`, `x`, `q`, `p`.

**Cel techniczny:** Przepisać pętlę z czytelnymi nazwami.

**Poziom trudności:** 1/5

- `s` może stać się `total_revenue`.

```python
orders_raw = [
    {"invoice_no": "536365", "stock_code": "85123A", "description": "WHITE HANGING HEART T-LIGHT HOLDER", "quantity": "6", "unit_price": "2.55", "customer_id": "17850", "country": "United Kingdom"},
    {"invoice_no": "536366", "stock_code": "71053", "description": "WHITE METAL LANTERN", "quantity": "-1", "unit_price": "3.39", "customer_id": "17850", "country": "United Kingdom"},
    {"invoice_no": "536367", "stock_code": "84406B", "description": "CREAM CUPID HEARTS COAT HANGER", "quantity": "8", "unit_price": "0", "customer_id": "17850", "country": "France"},
    {"invoice_no": "536368", "stock_code": "84029G", "description": "KNITTED UNION FLAG HOT WATER BOTTLE", "quantity": "3", "unit_price": "3.39", "customer_id": None, "country": "Germany"},
    {"invoice_no": "536369", "stock_code": "22752", "description": "SET 7 BABUSHKA NESTING BOXES", "quantity": "2", "unit_price": "7.65", "customer_id": "13047", "country": "United Kingdom"}
]

s = 0
c = 0
bad = 0

for x in orders_raw:
    q = int(x["quantity"])
    p = float(x["unit_price"])
    if q > 0 and p > 0:
        s = s + q * p
        c = c + 1
    else:
        bad = bad + 1

print("TOTAL", s)
print("VALID", c)
print("Errors", bad)


total_revenue = 0.0
valid_orders_count = 0
invalid_orders_count = 0

for order in orders_raw:
    quantity = int(x["quantity"])
    price = float(x["unit_price"])
    if q > 0 and p > 0:
        total_revenue = total_revenue + quantity * price
        valid_orders_count= valid_orders_count + 1
    else:
        invalid_orders_count = invalid_orders_count + 1

print("Total revenue:", total_revenue)
print("Valid orders:", valid_orders_count)
print("Invalid orders:", invalid_orders_count)
```

## Zadanie 7 — Funkcja `calculate_total_revenue`

**Opis biznesowy:** Zespół chce wielokrotnie liczyć przychód dla różnych zestawów zamówień.

**Cel techniczny:** Zamknąć logikę liczenia przychodu w funkcji.

**Poziom trudności:** 2/5

**Szacowany czas:** 20 min

**Wskazówki:**

- Funkcja powinna przyjmować listę zamówień.
- Funkcja powinna korzystać z `is_valid_order`.
- Funkcja powinna korzystać z `calculate_line_total`.
- Funkcja powinna zwracać liczbę, a nie drukować wynik.

```python
def calculate_total_revenue(orders: list) -> float:
    total_revenue = 0.0

    for order in orders:
        if is_valid_order(order):
            quantity = int(order["quantity"])
            unit_price = float(order["unit_price"])
            total_revenue += calculate_line_total(quantity, unit_price)

    return total_revenue


print(calculate_total_revenue(orders_raw))
```

## Zadanie 8 — Funkcja raportująca

## Zadanie 27 — Type hints dla funkcji obliczeniowych

**Opis biznesowy:** Funkcje liczące przychód powinny jasno mówić, jakich typów oczekują.

**Cel techniczny:** Dodać type hints do `calculate_line_total` i `calculate_total_revenue`.

**Poziom trudności:** 1/5

**Szacowany czas:** 15 min

**Wskazówki:**

- `quantity` to zwykle `int`.
- `unit_price` to zwykle `float`.
- Wynik obliczenia przychodu to `float`.

```python
type Order = dict[str, str]
type Orders = list[Order]

MIN_VALID_QUANTITY = 1
MIN_VALID_PRICE = 0.0
UNKNOWN_CUSTOMER_ID = "Unknown"

orders: Orders = [
    {"invoice_no": "536365", "quantity": "6", "unit_price": "2.55", "country": "United Kingdom"},
    {"invoice_no": "536366", "quantity": "-1", "unit_price": "3.39", "country": "United Kingdom"},
    {"invoice_no": "536367", "quantity": "8", "unit_price": "0", "country": "France"},
]

def typed_is_valid_quantity(quantity: int) -> bool:
     return quantity >= MIN_VALID_QUANTITY

def typed_is_valid_price(price: float) -> bool:
     return price > MIN_VALID_PRICE

def typed_is_valid_order(order: Order) -> bool:
     quantity = int(order["quantity"])
     price = float(order["unit_price"]) 
     return is_valid_quantity(quantity) and is_valid_price(price)


def typed_calculate_line_total(quantity: int, unit_price: float) -> float:
    return quantity * unit_price


def typed_calculate_total_revenue(orders: Orders) -> float:
    total_revenue = 0.0

    for order in orders:
        if typed_is_valid_order(order):
            quantity = int(order["quantity"])
            price = float(order["unit_price"])
            total_revenue += typed_calculate_line_total(quantity, price)

total_revenue = calculate_total_revenue(orders)
print(total_revenue)
```
