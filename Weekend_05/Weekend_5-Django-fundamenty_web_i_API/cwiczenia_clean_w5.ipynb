## Opis case'u biznesowego

Budujemy prosty backend dla aplikacji do zarządzania zadaniami w projekcie zespołowym.

Taki typ aplikacji jest podobny do uproszczonych systemów ticketowych, narzędzi typu Jira/Trello/Asana, wewnętrznych paneli operacyjnych oraz modułów zarządzania pracą w CRM-ach.

Na tym etapie nie skupiamy się jeszcze na rozbudowanej bazie danych. Najważniejsze jest zrozumienie przepływu request -> routing -> view -> serializer -> response.

## Opis danych

**Nazwa zbioru:** Task Management API — dane projektowe aplikacji kursowej

**Źródło:** dane tworzone lokalnie podczas ćwiczeń przez endpoint API. W realnej aplikacji analogiczne dane pochodziłyby z bazy PostgreSQL, systemu ticketowego, CRM-a albo narzędzia do zarządzania projektami.

**Główna encja:** `Task`

| Kolumna | Typ | Opis biznesowy |
|---|---|---|
| `id` | integer | Unikalny identyfikator zadania. Nadawany przez backend. |
| `title` | string | Krótka nazwa zadania widoczna na liście. Pole wymagane. |
| `description` | string | Dłuższy opis zadania. Pole opcjonalne. |
| `status` | string | Aktualny status pracy nad zadaniem. |
| `priority` | string | Priorytet zadania używany do planowania pracy. |
| `created_at` | datetime/string | Data utworzenia zadania. Na tym etapie może być pominięta albo dodana później. |

**Kategorie pola `status`:**

- `todo` — zadanie do zrobienia,
- `in_progress` — zadanie w trakcie realizacji,
- `done` — zadanie zakończone.

**Kategorie pola `priority`:**

- `low` — niski priorytet,
- `medium` — średni priorytet,
- `high` — wysoki priorytet.

**Przykładowy obiekt JSON:**

{
  "id": 1,
  "title": "Prepare Django project",
  "description": "Create initial backend structure",
  "status": "todo",
  "priority": "high"
}
