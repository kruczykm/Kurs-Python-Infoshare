# Część 1 — Bezpieczne odczytywanie plików i obsługa wyjątków

W tej części nauczysz się, jak otworzyć plik tekstowy w Pythonie i co zrobić, gdy pliku brakuje na dysku.

## Zadanie 1 — Odczyt zawartości pliku tekstowego

**Opis biznesowy:** Aplikacja do notatek musi potrafić wczytać całą treść pliku tekstowego i pokazać ją użytkownikowi.

**Cel techniczny:** Napisać funkcję `read_note_file(file_path: str | Path) -> str`, która otwiera plik z kodowaniem `utf-8` i zwraca jego zawartość.

**Poziom trudności:** 1/5 | **Szacowany czas:** 10 min

**Wskazówki:**
- Użyj menedżera kontekstu `with open(file_path, "r", encoding="utf-8") as f:`.
- Zwróć wynik metody `f.read()`.

```python
# Ścieżka do pliku z notatką
SAMPLE_NOTE = "./mini_cli/note_app_cli/notes_sample.txt"


def read_note_file(file_path: str) -> str:
        # Używamy zmiennej file_path, którą funkcja dostała w argumencie
        with open(file_path, "r", encoding="utf-8") as f:
            return f.read()  # Zwracamy treść na zewnątrz


# Test 1:
content = read_note_file(SAMPLE_NOTE)
print("--- Zawartość notatki ---")
print(content)
```

## Zadanie 2 — Obsługa braku pliku (`FileNotFoundError`)

**Opis biznesowy:** Jeśli użytkownik poda nazwę pliku, który nie istnieje, program nie powinien wyrzucać brzydkiego tracebacka, lecz wyświetlić czytelny komunikat.

**Cel techniczny:** Zmodyfikować funkcję `safe_read_note(file_path)` tak, aby łapała `FileNotFoundError` i zwracała informację o braku pliku.

**Poziom trudności:** 1/5 | **Szacowany czas:** 10 min

**Wskazówki:**
- Użyj bloku `try ... except FileNotFoundError:`.

```python
from pathlib import Path

# Ścieżka do pliku z notatką, lepiej jest uzywac Path
SAMPLE_NOTE = Path("./mini_cli/note_app_cli/notes_sample.txt")

def safe_read_note(file_path: Path) -> str | None:
    """Odczytuje plik i zwraca jego treść jako string. 
    W razie braku pliku zwraca None."""
    try:
        # Używamy zmiennej file_path, którą funkcja dostała w argumencie
        with open(file_path, "r", encoding="utf-8") as f:
            return f.read()  # Zwracamy treść na zewnątrz
    except FileNotFoundError:
        print(f"Błąd: Plik '{file_path}' nie istnieje!")
        return None


# Test 1:
content = safe_read_note(SAMPLE_NOTE)

if content:
    print("--- Zawartość notatki ---")
    print(content)
```

## Zadanie 3 — Dopisywanie nowej notatki do pliku (`append`)

**Opis biznesowy:** Użytkownik chce dopisać nową myśl lub zadanie do istniejącego pliku notatek bez kasowania dotychczasowej zawartości.

**Cel techniczny:** Napisać funkcję `append_note(file_path: str | Path, new_line: str) -> None`, która otwiera plik w trybie dopisywania (`"a"`).

**Poziom trudności:** 1/5 | **Szacowany czas:** 10 min

**Wskazówki:**
- Użyj trybu `"a"` (append): `with open(file_path, "a", encoding="utf-8") as f:`.
- Pamiętaj o dodaniu znaku nowej linii `\n` na końcu dopisywanego tekstu.

```python
from pathlib import Path

# Ścieżka do pliku z notatką, lepiej jest uzywac Path
SAMPLE_NOTE = Path("./mini_cli/note_app_cli/notes_sample.txt")

def read_note_file(file_path: str | Path) -> str:
    info_file = None
    """Odczytuje plik i zwraca jego treść jako string. 
    W razie braku pliku zwraca None."""
    try:
        # Używamy zmiennej file_path, którą funkcja dostała w argumencie
        with open(file_path, "r", encoding="utf-8") as f:
            info_file = f.read()  # Zwracamy treść na zewnątrz
            return info_file
    except FileNotFoundError:
        info_file = (f"Błąd: Plik '{file_path}' nie istnieje!")
        return info_file

def append_note(file_path: str | Path, new_line: str) -> bool:
    try:
        # Używamy zmiennej file_path, którą funkcja dostała w argumencie, "a" - append pozwoli nam dodac tresc do istniejacego pliku
        with open(file_path, "a", encoding="utf-8") as f:
            f.write(f"\n{new_line}")
            return True
    except Exception as e:
        print(f"Wystąpił błąd podczas zapisu: {e}")
        return False  # Zwracamy False, jeśli coś poszło nie tak

# Test 3:
is_done = append_note(SAMPLE_NOTE, "4. Przetestować mini-CLI.")
content = read_note_file(SAMPLE_NOTE)

if content and is_done:
    print("--- Zawartość notatki po zmianie ---")
    print(content)
```

# Część 2 — Budowa prostego mini-CLI za pomocą `sys.argv`

W klasycznym, początkującym podejściu do **mini-CLI** nie używamy jeszcze skomplikowanego modułu `argparse`. Korzystamy z wbudowanej listy `sys.argv`, która przechowuje argumenty przekazane z terminala (np. `python app.py read dziennik.txt`).

## Zadanie 4 — Analiza argumentów w `sys.argv`

**Opis biznesowy:** Nasze mini-CLI ma obsługiwać prostą komendę:
`python notes_cli.py read <plik>` albo `python notes_cli.py count <plik>`.

**Cel techniczny:** Napisać funkcję `handle_command(args: list[str]) -> str`, która sprawdza pierwszą akcję (`read` lub `count`) i wykonuje odpowiednią operację.

**Poziom trudności:** 2/5 | **Szacowany czas:** 20 min

**Wskazówki:**
- Jeśli `len(args) < 2`, zwróć komunikat o błędzie (np. `"Brak polecenia"`).
- Sprawdź `command = args[0]`. Jeśli `command == "read"`, zwróć treść pliku. Jeśli `command == "count"`, zwróć liczbę linijek w pliku.

```python
from pathlib import Path

# Ścieżka do pliku z notatką, lepiej jest uzywac Path
SAMPLE_NOTE = Path("./mini_cli/note_app_cli/notes_sample.txt")


def read_note_file(file_path: str | Path) -> str | None:
    info_file = None
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            info_file = f.read()  # Zwracamy treść na zewnątrz
            return info_file
    except FileNotFoundError:
        print(f"Błąd: Plik '{file_path}' nie istnieje!")
        return None


def count_lines(content_file: str) -> int:
    lines = content_file.splitlines() # Metoda splitlines() w Pythonie służy do dzielenia tekstu na poszczególne linijki i zwraca je w postaci listy stringów
    lines = len(lines) # metoda len() sprawdzamy ile mam elementow w liscie
    return lines


def handle_command(args: list[str]) -> str:
    info_message = None

    if len(args) < 2:
        info_message = "Błąd: Brak wymaganych argumentów."
        return info_message

    action = args[0]
    content = read_note_file(args[1])

    if content is None:
        return f"Nie można wykonać operacji z powodu błędu pliku."
    
    if args[0] == "read":
        return content
    elif args[0] == "count":
        info_message = f"Liczba linijek w pliku: {count_lines(content)}"
        return info_message
    else:
        info_message = f"Nieznana akcja: {action}"
        return info_message

# Test  4:
print(handle_command(["read", str(SAMPLE_NOTE)]))
print(handle_command(["count", str(SAMPLE_NOTE)]))
```

## Zadanie 5 — Mini-projekt: Kompletny skrypt `notes_cli.py`

**Opis biznesowy:** Złóżmy poznane klocki (odczyt plików, wyjątki, `sys.argv`) w jeden samodzielny skrypt terminalowy.

**Cel techniczny:** Zapisać plik `notes_cli.py` w katalogu roboczym i przetestować jego działanie za pomocą `subprocess.run()`.

```python
from pathlib import Path

# Definiujemy ścieżkę, pod którą zostanie utworzony gotowy skrypt CLI
CLI_SCRIPT = Path("./mini_cli/note_app_cli/notes_cli.py")

# W tej zmiennej przechowujemy cały kod źródłowy (jako tekst), 
# który zostanie zapisany w nowym pliku.
cli_code = '''
from pathlib import Path
import sys

def read_note_file(file_path: str | Path) -> str | None:
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Błąd: Plik '{file_path}' nie istnieje!")
        return None

def append_note(file_path: str | Path, new_line: str) -> bool:
    try:
        with open(file_path, "a", encoding="utf-8") as f:
            f.write(f"\\n{new_line}")
            return True
    except Exception as e:
        print(f"Wystąpił błąd podczas zapisu: {e}")
        return False

def count_lines(content_file: str) -> int:
    return len(content_file.splitlines())

def handle_command(args: list[str]) -> str:
    # Router komend: sprawdza, czy podano wystarczającą liczbę argumentów
    if len(args) < 2:
        return "Błąd: Brak wymaganych argumentów."

    action = args[0]   # Pierwszy argument to akcja (np. read, count, append)
    filepath = args[1] # Drugi argument to ścieżka do pliku

    # Obsługa specyficzna dla zapisu (wymaga dodatkowego argumentu z tekstem)
    if action == "append":
        if len(args) < 3:
            return "Błąd: Podaj tekst do dopisania!"
        text_to_add = args[2]
        success = append_note(filepath, text_to_add)
        if success:
            return "Notatka została pomyślnie dopisana!"
        return "Nie udało się dopisać notatki."

    # Dla operacji odczytu i zliczania najpierw pobieramy zawartość pliku
    content = read_note_file(filepath)
    if content is None:
        return "Nie można wykonać operacji z powodu błędu pliku."

    # Rozgałęzienie logiki na podstawie wybranej akcji
    if action == "read":
        return content
    elif action == "count":
        return f"Liczba linijek w pliku: {count_lines(content)}"
    else:
        return f"Nieznana akcja: {action}"

def main():
    # Pobieramy argumenty przekazane z terminala (odcinając nazwę samego skryptu)
    args = sys.argv[1:]
    
    # Jeśli użytkownik podał za mało argumentów, wyświetlamy pomoc i przerywamy działanie
    if len(args) < 2:
        print("Użycie:")
        print("  python notes_cli.py read <plik>")
        print("  python notes_cli.py count <plik>")
        print("  python notes_cli.py append <plik> \\"Tekst\\"")
        sys.exit(1)
        
    # Przekazujemy argumenty do funkcji zarządządzającej i wypisujemy wynik w konsoli
    result = handle_command(args)
    print(result)

# Punkt wejścia dla generowanego skryptu
if __name__ == "__main__":
    main()
'''

# Fizycznie zapisujemy przygotowany tekst jako plik .py na dysku
CLI_SCRIPT.write_text(cli_code.strip() + "\n", encoding="utf-8")

# Weryfikacja, czy plik został poprawnie utworzony
print("Sukces! Wygenerowano notes_cli.py z wykorzystaniem handle_command:", CLI_SCRIPT.exists())
```

## Zadanie 6 — Testowanie mini-CLI w akcji

Uruchommy nasz skrypt z poziomu notebooka za pomocą modułu `subprocess`, tak jakbyśmy wpisali polecenie w terminalu.

```python
import subprocess
import sys
from pathlib import Path

# Definiujemy ścieżkę do pliku testowego, na którym będziemy operować
SAMPLE_NOTE = Path("./mini_cli/note_app_cli/notes_sample.txt")

# Test 1: Testujemy komendę "read" (odczyt zawartości pliku)
res1 = subprocess.run(
    [sys.executable, "./mini_cli/note_app_cli/notes_cli.py", "read", str(SAMPLE_NOTE)], 
    capture_output=True,  # Przechwytuje to, co skrypt wypisuje na ekran (stdout) i błędy (stderr)
    text=True,            # Automatycznie zamienia bajty na ciąg znaków (string)
    encoding="utf-8"      # Wymusza kodowanie UTF-8, żeby polskie litery i komunikaty wyświetlały się poprawnie
)
# Wyświetlamy wynik działania pierwszej komendy
print(f"STDOUT (read):\n{res1.stdout}")

# Test 2: Testujemy komendę "count" (zliczanie linijek w pliku)
res2 = subprocess.run(
    [sys.executable, "./mini_cli/note_app_cli/notes_cli.py", "count", str(SAMPLE_NOTE)], 
    capture_output=True, 
    text=True, 
    encoding="utf-8"
)
# Wyświetlamy wynik zliczenia linijek
print(f"STDOUT (count):\n{res2.stdout}")

# Test 3: Testujemy zachowanie programu dla pliku, który NIE istnieje (sprawdzenie obsługi błędów)
res3 = subprocess.run(
    [sys.executable, "./mini_cli/note_app_cli/notes_cli.py", "read", "brak.txt"], 
    capture_output=True, 
    text=True, 
    encoding="utf-8"
)
# Wyświetlamy komunikat o błędzie (bierzemy stdout, a jeśli pusty – to stderr)
print(f"STDERR / STDOUT (brak pliku):\n{res3.stdout or res3.stderr}")

# Sprawdzamy kod wyjścia (returncode) – np. 0 oznacza sukces, a inne wartości (np. 1 lub 2) błąd
print("Kod wyjścia:", res3.returncode)
```
